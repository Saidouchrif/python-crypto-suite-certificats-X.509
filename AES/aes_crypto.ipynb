Chiffrement et déchiffrement AES (Fernet) d’un message.

```python
from cryptography.fernet import Fernet

#Generation et sauvegarde de la clé AES
key=Fernet.generate_key()
with open("secret.key","wb") as key_file:
    key_file.write(key)
print("Clé AES générée et sauvegardée dans 'secret.key'.")
frent = Fernet(key)
message = b"Devoir 3A LIA"

#Chiffrement du message
encrypted_message = frent.encrypt(message)
print("Message chiffré :", encrypted_message)

#Déchiffrement du message
decrypted_message = frent.decrypt(encrypted_message)
print("Message déchiffré :", decrypted_message.decode())
```
