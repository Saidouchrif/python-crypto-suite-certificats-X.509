Objectif : Signer un document avec HMAC-SHA256.

```python
#hmac_sign
import hmac
import hashlib
import base64
#charger le document
with open("message.txt", "rb") as f:
    message = f.read()
#cle secrete
secret_key = b'secret_key_123'
#creer le HMAC
hmac_object = hmac.new(secret_key, message, hashlib.sha256)
hmac_digest = hmac_object.digest()

#enregistrer le HMAC dans un fichier
with open("message.hmac", "wb") as f:
    f.write(base64.b64encode(hmac_digest))
print("HMAC signature created and saved to message.hmac")
print("HMAC:", base64.b64encode(hmac_digest).decode())
```
