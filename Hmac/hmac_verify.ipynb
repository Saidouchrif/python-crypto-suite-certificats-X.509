Objectif : Vérifier l’intégrité du document signé.

```python
import hmac
import hashlib
import base64

#charger le message depuis le fichier
with open("message.txt", "rb") as file:
    message = file.read()
#clé secrète partagée
secret_key = b'secret_key_123'
#charger la signature depuis le fichier
with open("message.hmac", "rb") as file:
    signature = base64.b64decode(file.read())

#calculer la signature HMAC du message
computed_hmac = hmac.new(secret_key, message, hashlib.sha256).digest()
#vérifier si la signature correspond
if hmac.compare_digest(computed_hmac, signature):
    print("Signature valide.")
else:
    print("Signature invalide.")
```
